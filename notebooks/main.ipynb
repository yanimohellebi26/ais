# Vue générale sur le dataset

```python
overview = pd.DataFrame({
    "Colonne": df.columns,
    "TypeDeDonnée": [df[col].dtype for col in df.columns],
    "NonNull": [df[col].count() for col in df.columns],
    "Manquantes": [df[col].isna().sum() for col in df.columns],
    "ValeursUniques": [df[col].nunique() for col in df.columns],
    "ExempleValeurs": [df[col].unique()[:5] for col in df.columns]
})
display(overview)
```
